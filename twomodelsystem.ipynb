```python
# Normalize Featuree
from sklearn.preprocessing import MinMaxScaler
scaler = MinMaxScaler()


X_train_scaled = scaler.fit_transform(X_train)
# X_val_scaled = scaler.transform(X_val)
X_test_scaled = scaler.transform(X_test)

y_train_scaled = scaler.fit_transform(y_train)
# y_val_scaled = scaler.transform(y_val)
y_test_scaled = scaler.transform(y_test)
```

```python
def create_sequences(x, y, time_step=1):
    X_seq, y_seq = [], []
    for i in range(time_step, len(x)):
        X_seq.append(x[i-time_step:i])  # Past `time_step` days of features
        y_seq.append(y[i])  # Corresponding target (Close price)
    return np.array(X_seq), np.array(y_seq)

time_step =100  # Use past 100 days to predict next day's Close price

# Prepare sequences for Train, Validation, and Test sets
X_train_seq, y_train_seq = create_sequences(X_train_scaled, y_train_scaled, time_step)
# X_val_seq, y_val_seq = create_sequences(X_val_scaled, y_val_scaled, time_step)
X_test_seq, y_test_seq = create_sequences(X_test_scaled, y_test_scaled, time_step)

# Print final dataset shapes
print(f"Train sequences: {X_train_seq.shape}, {y_train_seq.shape}")
# print(f"Validation sequences: {X_val_seq.shape}, {y_val_seq.shape}")
print(f"Test sequences: {X_test_seq.shape}, {y_test_seq.shape}")
```

# Creating Model using LSTM:

```python
from tensorflow.keras.callbacks import EarlyStopping,ReduceLROnPlateau
from tensorflow.keras.callbacks import ModelCheckpoint, EarlyStopping

early_stop=EarlyStopping(monitor="val_loss",patience=7,restore_best_weights=True)
reduce_lr = ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5, min_lr=0.0001)

#earlystopping decides self that for how much epoch should it run,here in terms of val_loss,it stops till it can minimize val_loss at optimum or goes constant

# Save weights only for best model ////
# checkpointer = ModelCheckpoint(
#     filepath = 'weights_best.hdf5', 
#     verbose = 2, 
#     save_best_only = True
# )
```

```python
import keras
from keras.models import Sequential
from keras.layers import LSTM, Dense, Dropout
from keras.optimizers import Adam


# Build the LSTM model
model_close = Sequential()

# LSTM layer (input shape = (time_step, number of features))
model_close.add(LSTM(units=64,activation='tanh',return_sequences=True, input_shape=(X_train_seq.shape[1], X_train_seq.shape[2])))
model_close.add(Dropout(0.3))  # Dropout to prevent overfitting

# Additional LSTM layer
model_close.add(LSTM(units=128, return_sequences=True))
model_close.add(Dropout(0.4))

model_close.add(LSTM(units=64))
model_close.add(Dropout(0.3))

# //
model_close.add(Dense(units=25, activation='relu'))

# Output layer
model_close.add(Dense(units=1,activation='linear'))  # Predicting the next 'Close' price

# Compile the model_close
# optimizer=Adam(learning_rate=0.2)
model_close.compile(optimizer=Adam(), loss='mean_squared_error')

# Summary of the model_close
model_close.summary()
```

```python
# Train the LSTM model
lstm_model_close=model_close.fit(
    X_train_seq, y_train_seq,
    validation_split=0.2,
    epochs=50,
    batch_size=16,
    callbacks=[early_stop,reduce_lr],
    verbose=1

)
```
